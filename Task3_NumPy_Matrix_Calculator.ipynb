```python

# QSkill Internship - Task 3
# Interactive Matrix Calculator using NumPy

import numpy as np

# --------------------------------------------------
# 1. FUNCTION TO ACCEPT A MATRIX FROM THE USER
# --------------------------------------------------

def input_matrix(matrix_name):
    print(f"\nEnter Matrix {matrix_name}")

    while True:
        try:
            rows = int(input("Number of rows: "))
            columns = int(input("Number of columns: "))

            if rows <= 0 or columns <= 0:
                print("Rows and columns must be positive integers.")
                continue

            break

        except ValueError:
            print("Please enter valid integers.")

    matrix_data = []

    for i in range(rows):
        while True:
            try:
                row_input = input(
                    f"Enter row {i + 1} ({columns} numbers, separated by spaces): "
                )

                row = list(map(float, row_input.split()))

                if len(row) != columns:
                    print(f"Please enter exactly {columns} numbers.")
                    continue

                matrix_data.append(row)
                break

            except ValueError:
                print("Invalid input. Enter numbers separated by spaces.")

    return np.array(matrix_data)


# --------------------------------------------------
# 2. INPUT TWO MATRICES
# --------------------------------------------------

print("=" * 45)
print("       NUMPY MATRIX CALCULATOR")
print("=" * 45)

A = input_matrix("A")
B = input_matrix("B")

print("\nMatrix A:")
print(A)

print("\nMatrix B:")
print(B)


# --------------------------------------------------
# 3. DISPLAY AVAILABLE OPERATIONS
# --------------------------------------------------

def display_menu():
    print("\n" + "=" * 45)
    print("           MATRIX OPERATIONS")
    print("=" * 45)
    print("1. Addition (A + B)")
    print("2. Subtraction (A - B)")
    print("3. Matrix Multiplication (A x B)")
    print("4. Transpose of Matrix A")
    print("5. Transpose of Matrix B")
    print("6. Determinant of Matrix A")
    print("7. Determinant of Matrix B")
    print("8. Display both matrices")
    print("9. Exit")
    print("=" * 45)


# --------------------------------------------------
# 4. PERFORM THE SELECTED OPERATION
# --------------------------------------------------

while True:
    display_menu()
    choice = input("Enter your choice (1-9): ").strip()

    if choice == "1":
        if A.shape == B.shape:
            print("\nResult: A + B")
            print(np.add(A, B))
        else:
            print(
                "\nError: Addition requires matrices "
                "with the same dimensions."
            )

    elif choice == "2":
        if A.shape == B.shape:
            print("\nResult: A - B")
            print(np.subtract(A, B))
        else:
            print(
                "\nError: Subtraction requires matrices "
                "with the same dimensions."
            )

    elif choice == "3":
        if A.shape[1] == B.shape[0]:
            print("\nResult: A x B")
            print(np.matmul(A, B))
        else:
            print(
                "\nError: Columns of Matrix A must equal "
                "rows of Matrix B."
            )

    elif choice == "4":
        print("\nTranspose of Matrix A:")
        print(np.transpose(A))

    elif choice == "5":
        print("\nTranspose of Matrix B:")
        print(np.transpose(B))

    elif choice == "6":
        if A.shape[0] == A.shape[1]:
            print("\nDeterminant of Matrix A:")
            print(np.linalg.det(A))
        else:
            print("\nError: Determinant requires a square matrix.")

    elif choice == "7":
        if B.shape[0] == B.shape[1]:
            print("\nDeterminant of Matrix B:")
            print(np.linalg.det(B))
        else:
            print("\nError: Determinant requires a square matrix.")

    elif choice == "8":
        print("\nMatrix A:")
        print(A)
        print("\nMatrix B:")
        print(B)

    elif choice == "9":
        print("\nThank you for using the Matrix Calculator!")
        print("Task 3 completed successfully.")
        break

    else:
        print("\nInvalid choice. Please select a number from 1 to 9.")
```
